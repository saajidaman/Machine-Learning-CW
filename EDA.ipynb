## 1. Read the CSV
Start with five columns so you can read a whole customer record across the screen. The full table stays in `customers`.

```python
import pandas as pd

customers = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")
print(f"{customers.shape[0]:,} customers and {customers.shape[1]} columns")
customers[["tenure", "Contract", "MonthlyCharges", "TotalCharges", "Churn"]].head()
```

## 2. Check the columns before using them
`TotalCharges` looks numeric, but a few cells contain spaces. Converting it to a number exposes those blanks as missing. Leave them missing for now: Notebook 2 learns what to fill them with from training data.

```python
customers["TotalCharges"] = pd.to_numeric(customers["TotalCharges"], errors="coerce")
missing_total = customers["TotalCharges"].isna()
print("Missing total charges:", missing_total.sum())
print("Tenure for those customers:", customers.loc[missing_total, "tenure"].unique())
print("Duplicate customer IDs:", customers["customerID"].duplicated().sum())
print("Duplicate rows:", customers.duplicated().sum())
pd.DataFrame({
    "Data type": customers.dtypes.astype(str),
    "Distinct values": customers.nunique(),
    "Missing values": customers.isna().sum(),
})
```

All 11 missing totals belong to customers with zero months of tenure. A blank tells us that the total was unrecorded; it does not prove that the bill was zero. The customer ID is useful for checking duplicates but has no useful meaning as a predictor.
